```python
# =====================================================
# Install Required Packages
# =====================================================
!pip install -q torch torchvision torchaudio
!pip install -q transformers diffusers accelerate sentencepiece safetensors
!pip install -q gTTS pillow requests

# =====================================================
# Import Libraries
# =====================================================
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from diffusers import StableDiffusionPipeline
from gtts import gTTS

# =====================================================
# Select Device
# =====================================================
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using Device:", device)

# =====================================================
# Topic
# =====================================================
topic = "The benefits of renewable energy"

# =====================================================
# TEXT GENERATION
# =====================================================
# =====================================================
# TEXT GENERATION
# =====================================================
print("\nLoading FLAN-T5 Model...")

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")

model = AutoModelForSeq2SeqLM.from_pretrained(
    "google/flan-t5-base"
)

prompt = f"Write a short, engaging paragraph about {topic}."

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=80
)

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("\nGenerated Text:\n")
print(generated_text)
# =====================================================
# IMAGE GENERATION
# =====================================================
print("\nLoading Stable Diffusion...")

dtype = torch.float16 if device == "cuda" else torch.float32

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=dtype
)

pipe = pipe.to(device)

image_prompt = f"An illustration of {topic}, digital art, highly detailed"

image = pipe(
    image_prompt,
    num_inference_steps=25
).images[0]

image.save("content_image.png")

print("Image saved as content_image.png")

# =====================================================
# AUDIO GENERATION
# =====================================================
tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("content_audio.mp3")

print("Audio saved as content_audio.mp3")

print("\nExperiment Completed Successfully!")
```
